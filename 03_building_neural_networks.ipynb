### __Implementing Forward Propagation__

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.layers import Dense, Input
from tensorflow.keras import Sequential
from tensorflow.keras.activations import sigmoid, linear, relu
from tensorflow.keras.losses import binary_crossentropy
```

### Creating Neural Networks Approch 1

```python
# method 1

# input
x = np.array([[200.0, 17.0]])

# layer1
layer1 = Dense(units=3, activation='sigmoid')
a1 = layer1(x)

# layer2
layer2 = Dense(units=1, activation='sigmoid')
a2 = layer2(a1)

print(f'a1: {a1}')
print(f'a2: {a2}')
```

# Creating Neural Network Approah 2

```python
# method 2

# input
x = np.array([[200.0, 17.0],
              [120.0, 5.0],
              [425.0, 20.0],
              [212.0, 18.0]])
y = np.array([1,0,0,1])

# layer 1 and layer 2
layer1 = Dense(units=3, activation='sigmoid')
layer2 = Dense(units=1, activation='sigmoid')

# model
model = Sequential([layer1, layer2])

# prepare the model for training
model.compile(
    loss='binary_crossentropy',
    optimizer='adam'
)

# Train the model
model.fit(x, y, epochs=100)

# new input 
x_new = np.array([[180.0, 16.5]])

# Make predictions
predictions = model.predict(x_new)

print(f'prediction: {predictions}')
```

```python
# method 3

# input
x = np.array([[200.0, 17.0],
              [120.0, 5.0],
              [425.0, 20.0],
              [212.0, 18.0]])

#Correct output
y = np.array([1,0,0,1])

# build the model
model = Sequential([
    Dense(units=3, activation='sigmoid'),
    Dense(units=1, activation='sigmoid')
])

# prepare the model for training
model.compile(
    loss='binary_crossentropy',
    optimizer='adam'
)

# Train the model
model.fit(x, y, epochs=100)

# new input 
x_new = np.array([[180.0, 16.5]])

# Make predictions
predictions = model.predict(x_new)

print(predictions)
```

# ___________________________________________________________________________________________________________
### __model.compile()__ 
prepares the model for training by telling it how to calculate errors and update its weights.

```python
model.compile(
    loss='binary_crossentropy',
    optimizer='adam'
)

""" 
telling the model:
"When you train, use this method to measure mistakes and this method to update your weights."
So:
loss → How do I measure my mistake?
optimizer → How do I change my weights to reduce the mistake?
"""
""" 
loss → measures how wrong the model is.
adam → decides how to update the weights to reduce that error.
"""
```

### __model.fit()__
is used to train the neural network using training data.

* model.fit() trains the model by using the training data and updating its weights and biases.

```python
model.fit(x, y, epochs=100)
""" 
An epoch means one complete pass through the entire training dataset.
e.g: The model sees all 1,000 examples once → 1 epoch.

Why 100 times?
Because the model usually doesn't learn everything from the data in one pass.

So 100 is not a fixed rule. It is just an example. We choose the number of epochs based on how well the model learns.
"""
```

### __model.predict()__
is used to make predictions on new input data after the model has been trained.

```python
prediction = model.predict(x_new)
```
