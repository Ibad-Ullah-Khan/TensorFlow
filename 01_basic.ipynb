## __TensorFlow__
TensorFlow is a Python library/framework used to build and train machine learning and neural network models.

```python
import tensorflow as tf
```

### __Keras__
Keras is a high-level API used to build neural networks easily.

```python
from tensorflow.keras.layers import Dense, Input
```

### __.keras__
Use Keras tools that are provided by TensorFlow.
### __.layers__
.layers means the part of Keras that contains different neural network layers.
### __Dense__
Dense creates a fully connected neural network layer.

```python
# Example
Dense(10, activation="relu")

""" 
10 → the layer has 10 neurons
relu → activation function used by those neurons
"""
```

### __Input__
Input tells the model what shape the input data has.

```python
# Example
Input(shape=(400,))

""" 
Input(shape=(400,))
Means:
The model expects 400 input features
"""

# Dense  → creates neurons/layer
# Input  → defines the input shape

""" 
Dense and Input are both layers but
Dense → a layer containing neurons that perform calculations.
Input → defines the shape/number of input features. It doesn't perform neural-network calculations.
"""
```

```python
from tensorflow.keras import Sequential
```

### __Sequential__

Sequential is used to build a neural network model by putting layers one after another.

```python
model = Sequential([
    Input(shape=(400,)),
    Dense(25, activation="sigmoid"),
    Dense(15, activation="sigmoid"),
    Dense(1, activation="sigmoid")
])

# Sequential = puts layers together to create a model.
```

### __BinaryCrossentropy__

BinaryCrossentropy is a loss function used for binary classification

It measures how wrong the model's prediction is.

```python
y_true = [1, 0, 1]
y_pred = [0.9, 0.2, 0.7]

loss_fn = BinaryCrossentropy()
loss = loss_fn(y_true, y_pred)

print(loss)

# BinaryCrossentropy is a Keras class that implements the binary cross-entropy mathematical function.
```
