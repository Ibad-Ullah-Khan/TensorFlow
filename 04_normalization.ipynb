### __Normalization__
Normalization is changing the scale of your input data so that different features are on a similar scale.

```python
import tensorflow as tf
import numpy as np


# ============================================================
# 1. Create some sample data
# ============================================================

# Each row = one training example
# Each column = one feature
#
# Feature 1 = house size
# Feature 2 = number of bedrooms

X_train = np.array([
    [1000, 2],
    [1500, 3],
    [2000, 4],
    [2500, 5],
    [3000, 6]
], dtype=np.float32)


# ============================================================
# 2. Create the Normalization layer
# ============================================================

normalizer = tf.keras.layers.Normalization(axis=-1)

# tf.keras
#     → TensorFlow's Keras API
#
# layers
#     → contains different neural-network layers
#
# Normalization
#     → a layer that normalizes input features
#
# axis=-1
#     → tells TensorFlow to calculate normalization
#       separately for each feature/column.
#
# So our data:
#
# [1000, 2]  ← feature 1 and feature 2
# [1500, 3]
# [2000, 4]
# ...
#
# Feature 1 is normalized separately from Feature 2.


# ============================================================
# 3. Adapt the normalization layer to our training data
# ============================================================

normalizer.adapt(X_train)

# adapt()
#     → looks at the training data and calculates
#       the statistics needed for normalization.
#
# In particular, it calculates the mean and variance
# of each feature.
#
# For example:
#
# Feature 1 → mean and variance
# Feature 2 → mean and variance
#
# IMPORTANT:
# adapt() does NOT train the neural network.
# It only calculates the statistics needed for normalization.


# ============================================================
# 4. Apply normalization to the data
# ============================================================

X_normalized = normalizer(X_train)

# normalizer(X_train)
#     → passes X_train through the normalization layer.
#
# The layer transforms the original values into
# normalized values.


# ============================================================
# 5. Print the original data
# ============================================================

print("Original data:")
print(X_train)


# ============================================================
# 6. Print the normalized data
# ============================================================

print("\nNormalized data:")
print(X_normalized.numpy())


# ============================================================
# 7. Check the mean of the normalized data
# ============================================================

print("\nMean of normalized data:")
print(tf.reduce_mean(X_normalized, axis=0).numpy())
```
