```python
import tensorflow as tf
from tensorflow.keras.layers import Dense, Input
```

```python
# Creating a single neuron for regression layer

linear_layer = tf.keras.layers.Dense(units=1, activation='linear',)
```

```python
# Let's examine the weights.
linear_layer.get_weights()
```

```python
# Giving input features to the neuron
x = tf.constant([[2.0, 3.0]])
output = linear_layer(x)
print(f"Linear Model Output: {output}")
print(f"Weights and bias: {linear_layer.get_weights()}")

""" 
We give the input to the linear layer. The layer uses its existing weight and bias to perform the calculation \(z=wx+b\), and the result is returned and stored as a Tensor.
"""
```
